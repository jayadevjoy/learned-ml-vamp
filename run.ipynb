#### Achievable Rate & NMSE

##### Configuration

```python
nfft = 512              # Number of FFT points
nsamp = 1000            # Number of samples per simulation
nitvamp = 2             # Number of iterations for VAMP estimator
retrain = True          # Set to True if the VAMP estimator is retrained for each SNR/INR pair
train_steps = 200      # Number of training steps for VAMP estimator
des_known = True        # Set to True if the desired signal bandwidth is known
quantize = True         # Set to True if quantization is applied

snr_values = np.arange(10, 21, 10)     # Desired SNR (SNR)
inr_values = np.arange(30, 81, 5)      # Interferer SNR (INR)
```

##### Run Simulation

```python
simplot = AllGridSim(snr_list=snr_values,
                     inr_list=inr_values,
                     nfft=nfft, 
                     nsamp=nsamp,
                     nitvamp=nitvamp,
                     train_steps=train_steps,
                     retrain=retrain,
                     des_known=des_known,
                     quantize=quantize)

simplot.run_sim()
```

```python
if quantize:
    simplot.save(f"../results/data/vamp_quant_iter_{nitvamp}_epoch_{train_steps}.csv")
else:
    simplot.save(f"../results/data/vamp_iter_{nitvamp}_epoch_{train_steps}.csv")
```
