#### Achievable Rate & NMSE

##### Configuration

```python
nfft = 512              # Number of FFT points
nsamp = 1000            # Number of samples per simulation
nitvamp = 1             # Number of iterations for VAMP estimator
retrain = True          # Set to True if the VAMP estimator is retrained for each SNR/INR pair
train_steps = 1000      # Number of training steps for VAMP estimator
des_known = True        # Set to True if the desired signal bandwidth is known
neural_update = False   # Set to False if the VAMP estimator uses fixed betas
quantize = False        # Set to True if quantization is applied

snr_values = np.arange(5, 11, 5)       # Desired SNR (SNR)
inr_values = np.arange(30, 71, 2.5)    # Interferer SNR (INR)
```

```python
quant_tag = "quant" if quantize else "unquant"
update_tag = "neural" if neural_update else "fixed"
out_path = f"results/data/vamp_{quant_tag}_{update_tag}_iter_{nitvamp}_epoch_{train_steps}.csv"
```

##### Run Simulation

```python
simplot = AllGridSim(snr_list=snr_values,
                     inr_list=inr_values,
                     nfft=nfft, 
                     nsamp=nsamp,
                     nitvamp=nitvamp,
                     train_steps=train_steps,
                     retrain=retrain,
                     des_known=des_known,
                     neural_update=neural_update,
                     quantize=quantize)

simplot.run_sim(out_path)
```
